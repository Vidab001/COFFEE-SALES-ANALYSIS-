```python
%%writefile app.py

import streamlit as st

st.title("📊 Social Media Engagement Calculator")

st.write("Calculate your social media engagement rate.")

platform = st.text_input("Platform")
likes = st.number_input("Likes", min_value=0, step=1)
comments = st.number_input("Comments", min_value=0, step=1)
shares = st.number_input("Shares", min_value=0, step=1)
views = st.number_input("Views", min_value=1, step=1)

if st.button("Calculate Engagement"):

    total_engagement = likes + comments + shares

    engagement_rate = (total_engagement / views) * 100

    st.subheader("📈 Results")

    st.write(f"Platform: {platform}")
    st.write(f"Total Engagement: {total_engagement:,}")
    st.write(f"Engagement Rate: {engagement_rate:.2f}%")

    st.subheader("Engagement Breakdown")

    chart_data = {
        "Likes": likes,
        "Comments": comments,
        "Shares": shares
    }

    st.bar_chart(chart_data)
```
